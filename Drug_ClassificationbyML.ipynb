Importing libraries for project

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder,StandardScaler
```

Loading the dataset

```python
Drug=pd.read_csv('Drug (1).csv')
Drug.head()
```

```python
Drug.describe()
```

Checking for any null value

```python
Drug.isnull().sum()
```

Data preprocessing

```python
numerical_features=['Age','Na','K']
categorical_features=['Sex','BP','Cholesterol']
```

```python
for i in categorical_features:
  le = LabelEncoder()
  Drug[i]=le.fit_transform(Drug[i])
```

```python
scaler=StandardScaler()
Drug[numerical_features]=scaler.fit_transform(Drug[numerical_features])
```

```python
Drug.head()
```

BUILDING THE MODEL

```python
from sklearn.model_selection  import train_test_split
```

```python
x=Drug.drop('Drug',axis=1)
y=Drug['Drug']
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.05,random_state=42)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
model=DecisionTreeClassifier(max_depth=4,random_state=42)
model.fit(X_train,y_train)
```

```python
y_pred= model.predict(X_test)
```

Checking the Accuracy

```python
from sklearn.metrics import accuracy_score
```

```python
acc=accuracy_score(y_test,y_pred)
acc
```

Applying the model on unknown data

```python
age=56
sex="M"
bp="LOW"
chol="HIGH"
na=0.1
k=0.045

sex_map={'M':1,'F':0}
bp_map={'LOW':1,'NORMAL':2,'HIGH':0}
chol_map={'NORMAL':1,'HIGH':0}

sex=sex_map[sex]
bp=bp_map[bp]
chol=chol_map[chol]
new_data=pd.DataFrame([[age,sex,bp,chol,na,k]],columns=['Age','Sex','BP','Cholesterol','Na','K'])
new_data[numerical_features]=scaler.transform(new_data[numerical_features])
new_data
prediction=model.predict(new_data)
print("Predicted Drug:",prediction[0])
```
