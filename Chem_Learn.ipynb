```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df=pd.read_csv('Alcohol_bp.csv')
df.tail()
```

```python
plt.figure(figsize=(10,5))
plt.scatter(df['MW'],df['bp'], label='Data Point')
plt.xlabel('MW')
plt.ylabel('Bp')
plt.title("Molecular weight Vs Boiling Point")
plt.grid(True)
plt.legend()
plt.show()
```

```python
import sklearn.model_selection
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score
```

```python
x =df[['MW','carbons','degree','aliphatic','avg_aryl_position','cyclic']]
y =df['bp']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
model = LinearRegression()
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
mse=mean_squared_error(y_test,y_pred)
r2=r2_score(y_test,y_pred)
print(y_pred)
print(f'Mean Squared Error: {mse}')
print(f'R-squared: {r2}')
```

```python
plt.figure(figsize=(10,5))
plt.scatter(x_test['MW'],y_pred, label='Data Point')
plt.xlabel('MW')
plt.ylabel('Bp')
plt.title("Molecular weight Vs Boiling Point")
plt.grid(True)
plt.legend()
plt.show()
```
